```python
import qrcode

def create_qr_code(data: str, filename: str = "qrcode.png"):
    # Initialize QR configuration
    qr = qrcode.QRCode(
        version=1,  # Matrix size: 1 is 21x21; automatically expands if data requires more space
        error_correction=qrcode.constants.ERROR_CORRECT_H,  # High error correction (~30% recovery capacity)
        box_size=10,  # Pixel size of each individual square
        border=4,     # Standard 4-box quiet zone margin
    )
    
    # Add payload (URL or plain text)
    qr.add_data(data)
    qr.make(fit=True)

    # Render image with customizable colors
    img = qr.make_image(fill_color="black", back_color="white")

    # Save output image
    img.save(filename)
    
    # Returning the image object auto-displays it inside Jupyter Notebooks
    return img

# Example execution with a URL
create_qr_code("https://github.com", "github_qr.png")
```
